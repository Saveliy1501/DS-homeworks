```python
#стратегия
from abc import ABC, abstractmethod


class TaxStrategy(ABC):
    @abstractmethod
    def calculate_tax(self, price: float) -> float:
        pass


class USAStrategy(TaxStrategy):
    def calculate_tax(self, price: float) -> float:
        return price * 1.48


class EURStrategy(TaxStrategy):
    def calculate_tax(self, price: float) -> float:
        return price * 1.20


class Without_Tax_Strategy(TaxStrategy):
    def calculate_tax(self, price: float) -> float:
        return price


class Product:
    def __init__(self, name: str, price: float):
        self.name = name
        self._price = price

    @property
    def price(self) -> float:
        return self._price

    @price.setter
    def price(self, value: float) -> float:
        self._price = value


class Order:
    def __init__(self, products: list[Product], strategy: TaxStrategy):
        self.products = products
        self.strategy = strategy

    def calculate_total_with_tax(self) -> float:
        return sum(self.strategy.calculate_tax(item.price) for item in self.products)

    def calculate_total_without_tax(self) -> float:
        return sum(item.price for item in self.products)

    def statistic(self):
        print(f"C налогом {self.calculate_total_with_tax()}")
        print(f"Без налогом {self.calculate_total_without_tax()}")
```

```python
products = [Product("мышка", 500), Product("клавиатура", 700), Product("ноутбук", 2000)]
my_order = Order(products, USAStrategy())
my_order.statistic()
```

```python
from abc import ABC, abstractmethod

class Computer:
    def __init__(self,cpu:str, storage : str, motherboard:str):
        self.cpu = cpu
        self.storage = storage
        self.motherboard = motherboard
     

class ComputerBuilder(ABC):
    def __init__(self,computer:Computer):
        self._computer = computer

    @property
    @abstractmethod
    def computer(self) -> Computer:
        pass

    @abstractmethod
    def add_CPU(self):
        pass

    @abstractmethod
    def add_storage(self):
        pass

    @abstractmethod
    def add_motherboard(self):
        pass

class OfficeBuilder(ComputerBuilder):
    def add_CPU(self):
        self._computer.cpu = "простой процессор"

    def add_storage(self):
        self._computer.storage = "просто память"

    def add_motherboard(self):
        self._computer.storage = "просто материнка"

class GamingBuilder(ComputerBuilder):
    def add_CPU(self):
        self._computer.cpu = "крутой процессор"

    def add_storage(self):
        self._computer.storage = "крутой память"

    def add_motherboard(self):
        self._computer.storage = "крутой материнка"

class WorkStationBuilder(ComputerBuilder):
    def add_CPU(self):
        self._computer.cpu = "супер крутой процессор"

    def add_storage(self):
        self._computer.storage = "супер крутой память"

    def add_motherboard(self):
        self._computer.storage = "супер крутой материнка"

class Diractor:
    def __init__(self, builder):
        self.builder = builder
    
    def make(self):
        self.builder.add_CPU()
        self.builder.add_storage()
        self.builder.add_motherboard()

    
    





    
```

```python
#адаптер
class Adapter(ABC):
    @abstractmethod
    def pay(self, price:float):
        pass

class Cash:
    def __init__(self,total):
        self.total = total
    
    def to_hends(self, price):
        return price
    
class Card:
    def __init__(self,total):
        self.total = total
    
    def to_account(self, price):
        return price*0.98

class CashAdapter(Adapter):
    def __init__(self, card:Cash):
        self.card = card
        
    def pay(self, price):
        print("Кэш адаптер")
        return self.card.to_hends(price)

class CardAdapter(Adapter):
    def __init__(self, card:Card):
        self.card = card
        
    def pay(self, price):
        print("Кард адаптер")
        return self.card.to_account(price)



```

```python
cash = Cash(10)
cashadapter = CashAdapter(cash)

def my_func(money:Adapter):
    print(money.pay(10))

my_func(cashadapter)
```
