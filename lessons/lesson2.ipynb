```python
#1
def reverse_string(s: str):
    result = ""
    i = len(s) - 1
    while i >= 0:
        result += s[i]
        i -= 1
    return result


reverse_string("Привет")
```

```python
# 2
def count_vowels_consonants(s):
    vowels = "aeiouауеёиэяюоы"
    s_lower = s.lower()
    vowels_count, consonants_count = 0, 0

    for char in s_lower:
        if char.isalpha():
            if char in vowels:
                vowels_count += 1
            else:
                consonants_count += 1

    return {"гласные": vowels_count, "согласные": consonants_count}


count_vowels_consonants("привет")
```

```python
#3
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]

is_palindrome("Шалаш")
```

```python
# 4
def unique_symbols(s):
    dictionary = dict.fromkeys(s)
    s = "".join(dictionary.keys())
    return s


unique_symbols("Шалаш привет как дела")
```

```python
# 5
def count_words(s):
    return len(s.split(" "))


count_words("Шалаш привет как дела")
```

```python
# 6
def Caesar(s: str, k: int):
    s = "".join([chr(ord(char) + k) for char in s])
    return s


Caesar("abc", 2)
```

```python
# 7
def capitalize_words(s: str):
    s = " ".join(item[0].capitalize() + item[1:] for item in s.split(" "))
    return s


count_words("Шалаш привет как дела")
```

```python
# 8
def pyromid(k):
    j = 1
    a = 1
    for i in range(k):
        print((k - a) * " " + "#" * j)
        j += 2
        a += 1


pyromid(10)
```

```python
# 9
def valid_brackes(s: str):
    stack = []
    pairs = {")": "(", "}": "{", "]": "["}

    for char in s:
        if char in "({[":
            stack.append(char)
        elif char in ")}]":
            if not stack or stack.pop() != pairs[char]:
                print("неправильно")
                return
    print("правильно" if not stack else "неправильно")


valid_brackes("}({[])")
```
