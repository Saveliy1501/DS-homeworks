```python
# Zadanie 1

import random
import string
```

```python
def generate_password(length):
    if length < 4:
        print("The password needs to be at least 4 letters long")
    lower = string.ascii_lowercase
    upper = string.ascii_uppercase
    digits = string.digits
    special = "!@#$%^&*"
    pass_chars = [random.choice(lower),random.choice(upper),random.choice(digits),random.choice(special)]
    all_chars = lower + upper + digits + special
    for i in range(length-4):
        pass_chars.append(random.choice(all_chars))
    random.shuffle(pass_chars)
    return ''.join(pass_chars)
```

```python
from collections import Counter
```

```python
# Zadanie 2

def words_count(s):
    s_split = s.split()
    count = len(s_split)
    count_unq = len(set(s_split))
    most_com = Counter(s_split).most_common(1)
    summ = 0
    for word in s_split:
        summ += len(word)
    avg = round(summ/count, 2)
    return f"Words: {count}, Unique words: {count_unq}, The most common: {most_com}, Average length: {avg}"
```

```python
s = "Mama Mia Amerikano Cappuchino Mamma Mia"
words_count(s)
```

```python
# Zadanie 3

def game():
    num = random.randint(1,100)
    attempts = 0
    print("Try to guess the number")
    while True:
        guess = int(input("Your number: "))
        attempts += 1
        if guess < num:
            print("Higher")
            continue
        elif guess > num:
            print("Lower")
            continue
        else:
            print(f"You guessed! Your attempts {attempts}")
            break
```

```python
game()
```

```python
# Zadanie 4

def timer():
    sec = int(input("Write seconds: "))
    while sec >= 0:
        mins = sec // 60
        secs = sec % 60
        print(f"{mins:02d}:{secs:02d}", end="\r")
        time.sleep(1)
        sec -= 1
    print("Time out!")
              
```

```python
timer()
```
