```python
#1
def roll(matrix):
    if not matrix:
        return []
    
    rolled = [row[::-1] for row in matrix[::-1]]
    return rolled
            


matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]


roll(matrix)
```

```python
# 2

def magic_square(matrix):
    if not matrix:
        return []

    n = len(matrix)

    for row in matrix:
        if len(row) != n:
            return False

    target = sum(matrix[0])

    for row in matrix:
        if sum(row) != target:
            return False

    for j in range(n):
        colum = [row[j] for row in matrix]
        if sum(colum) != target:
            return False

    main_diag = sum([matrix[i][i] for i in range(n)])
    if main_diag != target:
        return False

    second_diag = sum([matrix[n - i - 1][i] for i in range(n)])
    if second_diag != target:
        return False

    return True


matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
matrix_2 = [[2, 2, 2], [2, 2, 2], [2, 2, 2]]


magic_square(matrix_2)
```

```python
#3
def roll_slice(matrix , k):
    if not matrix:
        return []
    n = len(matrix)
    outside = []

    for j in range(1,n):
        outside.append(matrix[0][j])
    
    for j in range(1,n):
        outside.append(matrix[j][n-1])

    for j in range(n-2,-1,-1):
        outside.append(matrix[n-1][j])
    
    for j in range(n-2,-1,-1):
        outside.append(matrix[j][0])
    
    rotated = outside[-k:]+outside[:-k] 



    for j in range(1,n):
        matrix[0][j] = rotated.pop(0)

    
    for j in range(1,n):
        matrix[j][n-1] = rotated.pop(0)

    for j in range(n-2,-1,-1):
        matrix[n-1][j] = rotated.pop(0)
    
    for j in range(n-2,-1,-1):
        matrix[j][0] = rotated.pop(0)
    
    return matrix
    
          


matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]


roll_slice(matrix , 2)
```
