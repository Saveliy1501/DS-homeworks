```python
import math as m


# 1
def mathematic(x: float) -> float:
    y = m.pow(
        m.cos(m.exp(x))
        + (m.log(1 + x)) ** 2
        + m.sqrt(m.exp(m.cos(x)) + (m.sin(m.pi * x)) ** 2)
        + m.sqrt(1 / x)
        + m.cos(x**2),
        m.sin(x),
    )
    return y


x = 1.79
print(mathematic(x))
```

```python
# 2
def serch_substr(substr, st):
    st = st.lower()
    substr = substr.lower()
    if st.find(substr) == -1:
        print("Мимо!")
    else:
        print("Есть контакт!")


st = "привет Мир"
sub = "мир"
serch_substr(sub, st)
```

```python
# 3
from collections import Counter


def count_chars(st):
    st = st.replace(" ", "")
    chars_with_space = Counter(st).most_common(3)
    return [item[0] for item in chars_with_space]


count_chars("qqqwwe   ")
```

```python
# 4
def count_it(st):
    st = st.replace(" ", "")
    numbers = Counter(st).most_common(3)
    return {int(num): count for num, count in numbers}


count_it("1212131233418883333389999999999999988")
```

```python
# 5
def check_queen(a, b, c, d):
    if a < 1 or a > 8 or b < 1 or b > 8 or c < 1 or c > 8 or d < 1 or d > 8:
        return False
    if (a, b) == (c, d):
        return False
    return a == c or b == d or abs(a - c) == abs(b - d)


def check_horse(a, b, c, d):
    if a < 1 or a > 8 or b < 1 or b > 8 or c < 1 or c > 8 or d < 1 or d > 8:
        return False
    if (a, b) == (c, d):
        return False
    return (abs(a - c), abs(b - d)) in [(2, 1), (1, 2)]


a = int(input("Введите а"))
b = int(input("Введите b"))
c = int(input("Введите c"))
d = int(input("Введите d"))
print(check_queen(a, b, c, d))
print(check_horse(a, b, c, d))
```

```python
# доп таска
def is_horse_bit(a, b, c, d):
    if a < 1 or a > 8 or b < 1 or b > 8 or c < 1 or c > 8 or d < 1 or d > 8:
        return False
    if (a, b) == (c, d):
        return False
    if (abs(a - c), abs(b - d)) in [(2, 1), (1, 2)]:
        print("Бьет на первом ходу")
        return True
    elif (abs(a - c), abs(b - d)) in [
        (4, 2),
        (2, 4),
        (3, 3),
        (1, 1),
        (0, 2),
        (2, 0),
        (1, 3),
        (3, 1),
        (4, 1),
        (1, 4),
        (0, 4),
        (4, 0),
    ]:
        print("Бьет на втором ходу")
        return True
    else:
        return False


a = int(input("Введите а"))
b = int(input("Введите b"))
c = int(input("Введите c"))
d = int(input("Введите d"))
print(is_horse_bit(a, b, c, d))
```
