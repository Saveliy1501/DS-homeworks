```python
import numpy as np
import pandas as pd



data = pd.read_csv('../lessons/train.csv')

features = ['OverallQual', 'GrLivArea', 'TotalBsmtSF', 'YearBuilt', 'GarageCars']
X = data[features].values
y = data['SalePrice'].values
```

```python

for i in range(X.shape[1]):
    X[np.isnan(X[:, i]), i] = np.nanmean(X[:, i])

mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
std[std == 0] = 1
X = (X - mean) / std

y_mean = np.mean(y)
y_std = np.std(y)
y = (y - y_mean) / y_std

X = np.hstack((np.ones((X.shape[0], 1)), X))
```

```python

np.random.seed(42)
indices = np.random.permutation(len(X))
split = int(0.8 * len(X))

X_train = X[indices[:split]]
X_test = X[indices[split:]]
y_train = y[indices[:split]]
y_test = y[indices[split:]]
print(f"Обучающая выборка: {X_train.shape[0]}, Тестовая: {X_test.shape[0]}")
```

```python
weights = np.zeros(X_train.shape[1])
learning_rate = 0.1
n_iterations = 1000
m = len(X_train)

for i in range(n_iterations):
    y_pred = np.dot(X_train, weights)
    error = y_pred - y_train
    loss = np.mean(error ** 2)
    gradient = (2 / m) * np.dot(X_train.T, error)
    weights = weights - learning_rate * gradient
    
    if i % 100 == 0:
        print(f"Итерация {i:4d} | Потери: {loss:.6f}")

print("\nВеса моделей:")
print(f"Смещение: {weights[0]:.4f}")
for i in range(1, len(weights)):
    print(f"{features[i-1]}: {weights[i]:.4f}")

y_train_pred = np.dot(X_train, weights)
y_test_pred = np.dot(X_test, weights)

y_train_orig = y_train * y_std + y_mean
y_test_orig = y_test * y_std + y_mean
y_train_pred_orig = y_train_pred * y_std + y_mean
y_test_pred_orig = y_test_pred * y_std + y_mean




print("\nПримеры")
print("Номер | Реальная   | Предсказанная | Ошибка")


for i in range(min(10, len(y_test_orig))):
    real = y_test_orig[i]
    pred = y_test_pred_orig[i]
    err = abs(real - pred)
    print(f"{i+1} | ${real:10,.0f} | ${pred:10,.0f} | ${err:8,.0f}")
```
