```python
#фабричный метод
from abc import ABC, abstractmethod

class Transport(ABC):
    @abstractmethod
    def deliver(self) -> str:
        pass

class Truck(Transport):
    def deliver(self) -> str:
        return "Доставка по земле (грузовик)"

class Ship(Transport):
    def deliver(self) -> str:
        return "Доставка по морю (корабль)"

class Plane(Transport):
    def deliver(self) -> str:
        return "Доставка по воздуху (самолёт)"

class Logistics(ABC):
    @abstractmethod
    def create_transport(self) -> Transport:
        pass

    def plan_delivery(self) -> str:
        transport = self.create_transport()
        return transport.deliver()

class RoadLogistics(Logistics):
    def create_transport(self) -> Transport:
        return Truck()

class SeaLogistics(Logistics):
    def create_transport(self) -> Transport:
        return Ship()

class AirLogistics(Logistics):
    def create_transport(self) -> Transport:
        return Plane()

 
```

```python
logistics = RoadLogistics()
print(logistics.plan_delivery()) 

logistics = SeaLogistics()
print(logistics.plan_delivery())  

logistics = AirLogistics()
print(logistics.plan_delivery())  
```

```python
#абстрактная фабрика
class Button(ABC):
    @abstractmethod
    def render(self) -> str:
        pass

class CheckBox(ABC):
    @abstractmethod
    def render(self) -> str:
        pass

class WinButton(Button):
    def render(self) -> str:
        return "Отрисована виндовс кнопка"

class MacButton(Button):
    def render(self) -> str:
        return "Отрисована мак кнопка"

class WinCheckBox(Button):
    def render(self) -> str:
        return "Отрисована виндовс чек бокс"

class MacCheckBox(Button):
    def render(self) -> str:
        return "Отрисована мак чек бокс"

class GUIFactory(ABC):
    @abstractmethod
    def create_button(self) -> Button:
        pass

    @abstractmethod
    def create_checkbox(self) -> CheckBox:
        pass

class WinFactoru(GUIFactory):
    def create_button(self) -> Button:
        return WinButton()
    
    def create_checkbox(self) -> CheckBox:
        return WinCheckBox()
    
class MacFactoru(GUIFactory):
    def create_button(self) -> Button:
        return MacButton()
    
    def create_checkbox(self) -> CheckBox:
        return MacCheckBox()


```

```python
def client(factory: GUIFactory):
    button = factory.create_button()
    checkbox = factory.create_checkbox()
    print(button.render())
    print(checkbox.render())

platform = input()
if platform == "windows":
    factory = WinFactoru()
else:
    factory = MacFactoru()
client(factory)
```

```python
#декоратор
class Employee(ABC):
    @abstractmethod
    def get_salary(self) -> float:
        pass


class BaseEmployee(Employee):
    def __init__(self, base_salary: float):
        self.base_salary = base_salary

    def get_salary(self) -> float:
        return self.base_salary


class BonusDecorator(Employee):
    def __init__(self, employee: Employee):
        self._employee = employee

    @abstractmethod
    def get_salary(self) -> float:
        pass


class ProductivityBonus(BonusDecorator):
    def get_salary(self) -> float:
        return self._employee.get_salary() * 1.20  


class SeniorityBonus(BonusDecorator):
    def get_salary(self) -> float:
        return self._employee.get_salary() + 5000 


class LeadershipBonus(BonusDecorator):
    def get_salary(self) -> float:
        return self._employee.get_salary() + 10000  

    
```

```python
emp = BaseEmployee(50000)
print(f"Базовая: {emp.get_salary()}")
emp_with_bonuses = ProductivityBonus(emp)
emp_with_bonuses = SeniorityBonus(emp_with_bonuses)
emp_with_bonuses = LeadershipBonus(emp_with_bonuses)

print(f"Итоговая: {emp_with_bonuses.get_salary()}")
```

```python
#одиночка
import logging
from logging.handlers import TimedRotatingFileHandler
from inspect import stack

class ContextLogger:
    _instance = None

    def __new__(cls, log_file="app.log"):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance._initialize(log_file)
        return cls._instance

    def _initialize(self, log_file):
        self.logger = logging.getLogger("AppLogger")
        self.logger.setLevel(logging.DEBUG)

        formatter = logging.Formatter(
            '%(asctime)s - %(levelname)s - [%(filename)s:%(lineno)d - %(funcName)s()] - %(message)s'
        )

        file_handler = TimedRotatingFileHandler(
            log_file, when="midnight", interval=1, backupCount=7, encoding="utf-8"
        )
        file_handler.setFormatter(formatter)
        file_handler.setLevel(logging.DEBUG)

        self.logger.addHandler(file_handler)

    def _get_context(self):
        frame = stack()[3] 
        module = frame.filename
        lineno = frame.lineno
        func = frame.function
        return module, func, lineno

    def debug(self, message):
        mod, func, line = self._get_context()
        self.logger.debug(f"[{mod}:{line} {func}()] {message}")

    def info(self, message):
        mod, func, line = self._get_context()
        self.logger.info(f"[{mod}:{line} {func}()] {message}")

    def error(self, message):
        mod, func, line = self._get_context()
        self.logger.error(f"[{mod}:{line} {func}()] {message}")



```

```python
logger1 = ContextLogger("my_app.log")
logger2 = ContextLogger("my_app.log")

print(logger1 is logger2)  

def test_function():
    logger1.debug("Отладочное сообщение")
    logger1.info("Информационное сообщение")
    logger1.error("Ошибка!")

test_function()
```
