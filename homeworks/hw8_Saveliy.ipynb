```python
def matsum_vector_scalar(vector, scalar):
    return [v + scalar for v in vector]
```

```python
print(f"{matsum_vector_scalar(simple_vector, scalar_value)} = {simple_vector + scalar_value}")
```

```python
def matsum_matrix_scalar(matrix, scalar):
    return [[v + scalar for v in row] for row in matrix]
```

```python
print(f"simple python: {matsum_matrix_scalar(simple_matrix, scalar_value)}\nnumpy: {(simple_matrix + scalar_value).tolist()}")
```

```python
def matmul_vector_vector(vector_first, vector_second):
    result = 0
    for v_f, v_s in zip(vector_first, vector_second):
        result += v_f * v_s
    return result
```

```python
#print(f"{matmul_vector_vector(simple_vector, simple_vector)} = {np.dot(simple_vector, simple_vector)}")
```

```python
def matmul_vector_matrix(matrix, vector):
    result = []
    for row in matrix:
        tmp = 0
        for m_v, v in zip(row, vector):
            tmp += m_v * v
        result.append(tmp)
    return result
```

```python
print(f"{matmul_vector_matrix(simple_matrix, simple_vector)} = {np.dot(simple_matrix, simple_vector)}")
```

**Лекарство от скуки _(домашка)_**

Реализовать оставшиеся методы (сложение матриц, перемножение матриц, транспонирование)

```python
def matrix_add(A, B):
    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] + B[i][j])
        result.append(row)
    return np.array(result)

def matrix_sub(A, B):  
    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] - B[i][j])
        result.append(row)
    return np.array(result)

def matrix_mul(A, B):
    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(A[0])):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)
    return np.array(result)

def matrix_transpose(A):
    result = []
    for j in range(len(A[0])):
        row = []
        for i in range(len(A)):
            row.append(A[i][j])
        result.append(row)
    return np.array(result)
```

```python
# Вариант 1
A = np.array([[1,2,4],[7,5,-2],[-2,1,3]])
B = np.array([[-1,0,3],[2,-6,1],[-7,1,5]])
C = np.array([[-2,3],[4,0],[7,-1]])

print("A+B:\n", matrix_add(A,B))
print("\nB-A:\n", matrix_sub(B,A))
print("\nA*C:\n", matrix_mul(A,C))
print("\nA*B*C:\n", matrix_mul(matrix_mul(A,B),C))
```
