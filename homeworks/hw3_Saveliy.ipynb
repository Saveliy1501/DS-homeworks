```python
# 3
def create_list(n: int):
    return [i for i in range(n, 0, -1)]


create_list(5)
```

```python
# 4
def centuary(year: int):
    if year < 100: return 1
    return int(str(year)[:-2]) + 1

centuary(2026)
```

```python
#5
def reverse_int(num:int):
    return [int(item) for item in str(num)[::-1]]

reverse_int(12345)
```

```python
# 6
def horisontal_round(n: int):
    if n < 0:
        return []
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    current = 1

    for i in range(n):
        for j in range(n):
            matrix[i][j] = current
            current += 1

    for i in range(1, n, 2):
        matrix[i] = matrix[i][::-1]
    return matrix


def virtical_round(n: int):
    if n < 0:
        return []
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    current = 1

    for i in range(n):
        for j in range(n):
            matrix[j][i] = current
            current += 1

    for i in range(1, n, 2):
        colum = [item[i] for item in matrix][::-1]
        for j in range(n):
            matrix[j][i] = colum[j]

    return matrix


print(horisontal_round(6))
virtical_round(6)
```
