```python
import pandas as pd

class CrashAnalyser:

    def __init__(self, filename="train_data.csv"):
        self.filename = filename

    def analyse_survived(self):
        df = pd.read_csv(self.filename)
        survived_people = df[df["Survived"] == 1]
        killed_people = df[df["Survived"] == 0]
        

        common_sex = df["Sex"].mean()
        sex_survived = survived_people["Sex"].mean()
        sex_killed = killed_people["Sex"].mean()
        print("1. ПОЛ :")
        print(f"Среднее по всем: {common_sex:.3f}")
        print(f"У выживших: {sex_survived:.3f}")
        print(f"У погибших: {sex_killed:.3f}")
        if common_sex < sex_survived:
            print("Женщины погибали чаще")
        else:
            print("Мужчины погибали чаще")
        
        common_age = df["Age"].mean()
        age_survived = survived_people["Age"].mean()
        age_killed = killed_people["Age"].mean()
        print("2. ВОЗРАСТ:")
        print(f"Средний по всем: {common_age:.3f}")
        print(f"У выживших: {age_survived:.3f}")
        print(f"У погибших: {age_killed:.3f}")
        if age_survived < age_killed:
            print("Выжившие в среднем моложе")
        else:
            print("Выжившие в среднем старше")
        
        common_fare = df["Fare"].mean()
        fare_survived = survived_people["Fare"].mean()
        fare_killed = killed_people["Fare"].mean()
        print("3. ЦЕНА БИЛЕТА:")
        print(f"Средняя по всем: {common_fare:.3f}")
        print(f"У выживших: {fare_survived:.3f}")
        print(f"У погибших: {fare_killed:.3f}")
        if fare_survived > fare_killed:
            print("У выживших билеты дороже")
        else:
            print("У погибших билеты дороже")
        
        common_family = df["Family_size"].mean()
        family_survived = survived_people["Family_size"].mean()
        family_killed = killed_people["Family_size"].mean()
        print("4. РАЗМЕР СЕМЬИ:")
        print(f"Средний по всем: {common_family:.3f}")
        print(f"У выживших: {family_survived:.3f}")
        print(f"У погибших: {family_killed:.3f}")
        if family_survived > family_killed:
            print("У выживших семьи больше")
        else:
            print("У погибших семьи больше")
        
        print("5. КЛАСС:")
        pclass_1_total = df["Pclass_1"].sum()
        pclass_1_killed = killed_people["Pclass_1"].sum()
        pclass_1_death_rate = (pclass_1_killed / pclass_1_total * 100) if pclass_1_total > 0 else 0
        print(f"Первый класс: погибло {pclass_1_death_rate:.1f}% пассажиров")
        
        pclass_2_total = df["Pclass_2"].sum()
        pclass_2_killed = killed_people["Pclass_2"].sum()
        pclass_2_death_rate = (pclass_2_killed / pclass_2_total * 100) if pclass_2_total > 0 else 0
        print(f"Второй класс: погибло {pclass_2_death_rate:.1f}% пассажиров")
        
        pclass_3_total = df["Pclass_3"].sum()
        pclass_3_killed = killed_people["Pclass_3"].sum()
        pclass_3_death_rate = (pclass_3_killed / pclass_3_total * 100) if pclass_3_total > 0 else 0
        print(f"Третий класс: погибло {pclass_3_death_rate:.1f}% пассажиров")
        
        print("6. ТИТУЛ:")
        title_1_total = df["Title_1"].sum()
        title_1_killed = killed_people["Title_1"].sum()
        title_1_death_rate = (title_1_killed / title_1_total * 100) if title_1_total > 0 else 0
        print(f"Title_1: погибло {title_1_death_rate:.1f}% пассажиров")
        
        title_2_total = df["Title_2"].sum()
        title_2_killed = killed_people["Title_2"].sum()
        title_2_death_rate = (title_2_killed / title_2_total * 100) if title_2_total > 0 else 0
        print(f"Title_2: погибло {title_2_death_rate:.1f}% пассажиров")
        
        title_3_total = df["Title_3"].sum()
        title_3_killed = killed_people["Title_3"].sum()
        title_3_death_rate = (title_3_killed / title_3_total * 100) if title_3_total > 0 else 0
        print(f"Title_3: погибло {title_3_death_rate:.1f}% пассажиров")
        
        title_4_total = df["Title_4"].sum()
        title_4_killed = killed_people["Title_4"].sum()
        title_4_death_rate = (title_4_killed / title_4_total * 100) if title_4_total > 0 else 0
        print(f"Title_4: погибло {title_4_death_rate:.1f}% пассажиров")
        
        print("7. ПОРТ ПОСАДКИ:")
        emb_1_total = df["Emb_1"].sum()
        emb_1_killed = killed_people["Emb_1"].sum()
        emb_1_death_rate = (emb_1_killed / emb_1_total * 100) if emb_1_total > 0 else 0
        print(f"Emb_1: погибло {emb_1_death_rate:.1f}% пассажиров")
        
        emb_2_total = df["Emb_2"].sum()
        emb_2_killed = killed_people["Emb_2"].sum()
        emb_2_death_rate = (emb_2_killed / emb_2_total * 100) if emb_2_total > 0 else 0
        print(f"Emb_2: погибло {emb_2_death_rate:.1f}% пассажиров")
        
        emb_3_total = df["Emb_3"].sum()
        emb_3_killed = killed_people["Emb_3"].sum()
        emb_3_death_rate = (emb_3_killed / emb_3_total * 100) if emb_3_total > 0 else 0
        print(f"Emb_3: погибло {emb_3_death_rate:.1f}% пассажиров")
        


```

```python
a = CrashAnalyser()
a.analyse_survived()
```
