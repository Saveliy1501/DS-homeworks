```python
# 1
from math import sqrt


def quadratic_equation(a: float, b: float, c: float):
    D = b**2 - 4 * a * c
    if D > 0:
        return [(-b + sqrt(D)) / 2 * a, (-b - sqrt(D)) / 2 * a]
    elif D == 0:
        return [-b / 2 * a , None]
    else:
        print("Нет корней")
        return None


print(quadratic_equation(1, -12, 20))
print(quadratic_equation(1, 17, 72))
print(quadratic_equation(1, -7, -44))
print(quadratic_equation(1, 9, 8))
print(quadratic_equation(1, -2, -63))
for i in range(len(quadratic_equation(1, 4, -5))):
    print(quadratic_equation(1, 0, -8 + quadratic_equation(1, 4, -5)[i]))
```

```python
# 5
def all_eq(lst: list):
    maximum = max(len(item)for item in lst)
    return [item + (maximum - len(item)) * "_" for item in lst]


all_eq(["Hello", "hi", "are", "youyou", "a"])
```

```python
# 1
def run_lenth_encoding(st: str):
    if not st:
        return st
    reslt_list = []
    count = 1
    for i in range(1, len(st)):
        if st[i] == st[i - 1]:
            count += 1
        else:
            reslt_list.append(st[i - 1] + str(count))
            count = 1

    reslt_list.append(st[-1] + str(count))
    result = "".join(reslt_list)
    return result if len(result) < len(st) else st


run_lenth_encoding("aaabbcccc")
```

```python
#2
def is_one_step(large: str, small: str) -> bool:
    len_large, len_small = len(large), len(small)
    if len_small > len_large:
        return is_one_step(small, large)
    if len_large - len_small > 1:
        return False

    for i in range(len_small):
        if small[i] != large[i]:
            if len_small == len_large:
                return small[i + 1 :] == large[i + 1 :]
            else:
                return small[i:] == large[i + 1 :]

    return len_small + 1 == len_large

is_one_step("Hellow","Helrow")
```

```python
# 4
def is_match(text: str, pattern: str) -> bool:
    if not pattern:
        return not text
    
    first_match = bool(text) and (pattern[0] == text[0] or pattern[0] == '.')

    if len(pattern) >= 2 and pattern[1] == '*':
     
        return (is_match(text, pattern[2:]) or 
                (first_match and is_match(text[1:], pattern)))
    else:
      
        return first_match and is_match(text[1:], pattern[1:])


is_match("abc","a*bc")               
```

```python
#6
def tournament(arr, comparisons=0):
    if len(arr) == 1:
        return arr[0], comparisons
    winners = []
    for i in range(0, len(arr) - 1, 2):
        comparisons += 1
        winners.append(max(arr[i], arr[i + 1]))
    
    if len(arr) % 2:
        winners.append(arr[-1])
    
    return tournament(winners, comparisons)


print(tournament([5, 3, 8, 1, 9, 2]))  
    
```

```python
# 7
def wave_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        if i % 2 == 0:
            if arr[i] < arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
        else:
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
    return arr


wave_sort([1, 2, 3, 4, 5, 6, 7])
```

```python
#8
def spiral(matrix):
    if not matrix:
        return []
    
    result = []
    top = 0                    
    bottom = len(matrix) - 1   
    left = 0                   
    right = len(matrix[0]) - 1 
    
    while top <= bottom and left <= right:
        for col in range(left, right + 1):
            result.append(matrix[top][col])
        top += 1  
        
        for row in range(top, bottom + 1):
            result.append(matrix[row][right])
        right -= 1  

        if top <= bottom:
            for col in range(right, left - 1, -1):
                result.append(matrix[bottom][col])
            bottom -= 1  
        
        if left <= right:
            for row in range(bottom, top - 1, -1):
                result.append(matrix[row][left])
            left += 1  
    
    return result

matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

spiral(matrix)
```

```python
#9
def intersect_with_duplicates(nums1, nums2):
    nums1.sort()
    nums2.sort()

    i = j = 0
    result = []

    while i < len(nums1) and j < len(nums2):
        if nums1[i] < nums2[j]:
            i += 1
        elif nums1[i] > nums2[j]:
            j += 1
        else:
            result.append(nums1[i])
            i += 1
            j += 1
    return result


intersect_with_duplicates([1, 2, 2, 3], [2, 2, 3, 4])
```

```python
#10
def count_inversions(arr):
    n = len(arr)
    a = arr[:]
    inv_count = 0

    for i in range(n - 1):
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                inv_count += 1   
    return inv_count

count_inversions([1,2,4,3])
```
